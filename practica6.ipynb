```python
# Requests sirve para enviar solicitudes HTTP a servidores web, permite interactuar con páginas web.
import requests

# BeautifulSoup sirve para analizar y extraer datos de documentos HTML y XML.
from bs4 import BeautifulSoup

import pandas as pd
```

```python
# 1. Hacer la petición a la página objetivo
url = "https://www.taf.com.mx/asics"

```

```python
import requests
#Envía una petición GET al servidor. El servidor entrega todo el HTML de la página.
response = requests.get(url)
```

```python
# Garantiza que caracteres especiales se interpreten sin errores.
response.encoding = 'utf-8'
```

```python
# 2. Convertir el texto HTML a un objeto parseable con BeautifulSoup
# Toma el texto HTML de la respuesta y lo analiza con el motor "html.parser".
# Un objeto parseable es un dato en formato de texto plano que tiene la estructura correcta para ser analizado
texto = BeautifulSoup(response.text, "html.parser")
texto
```

```python
tenis = texto.find_all("article", class_='vtex-product-summary-2-x-element')
tenis

```

```python
datos = []

# 4. Iterar sobre cada tarjeta para "pescar" los datos sueltos
for articulo in tenis:
    # Extraer el título del atributo 'title' dentro de la etiqueta <a>
    nombre = articulo.find("div", class_="vtex-flex-layout-0-x-flexRow--ProductItemName").text

    # Extraer el precio dentro de la clase .price_color
    precio_prev = articulo.find("div", class_="vtex-flex-layout-0-x-flexRow--ProductItem-Price").text
    precio = float(precio_prev.replace("$", "").replace(",", ""))



    # Extraer la disponibilidad
    genero = articulo.find("div", class_="vtex-flex-layout-0-x-flexRow--ProductBrand").text

    datos.append({
        "nombre": nombre,
        "precio": precio,
        "genero": genero
    })
```

```python
df = pd.DataFrame(datos)
df
```

```python
df.to_csv("catalogo_tenis.csv", index=False)
```
